```python
import os
import sys
from contextlib import redirect_stdout

from src.pathgen import get_projz_file_path
from src.common.conllu_io import sentence_add_to_train, count_non_projectivity, count_deprels
```

```python
def get_splits(lang):
    raw_train_path, projz_train_path = get_projz_file_path(lang, "train")
    raw_dev_path, projz_dev_path = get_projz_file_path(lang, "dev")
    raw_test_path, projz_test_path = get_projz_file_path(lang, "test")
    return {
        "raw_train_path": raw_train_path, 
        "raw_dev_path": raw_dev_path,
        "raw_test_path": raw_test_path,
        "projz_train_path": projz_train_path,
        "projz_dev_path": projz_dev_path,
        "projz_test_path": projz_test_path,
    }


```

```python
langs = ["Uyghur"]
for lang in langs:
    with open(os.devnull, "w") as f, redirect_stdout(f):
        splits = get_splits(lang)
        train = splits["raw_train_path"]
        dev = splits["raw_dev_path"]
        test = splits["raw_test_path"]
    print(f"Treebank: {lang}")
    for split, split_name in [(train, "train"), (dev, "dev"), (test, "test")]:
        print(f"Split: {split_name}")
        non_proj_arcs, all_arcs, percent = count_non_projectivity(split)
        print(f"Count of non-projective arcs: {non_proj_arcs}")
        print(f"Count of all arcs: {all_arcs}")
        print(f"Percentage of non-projective arcs: {percent:.3f}%")
    print("---------------------------------------------------")
```

```python
langs = ["Uyghur"]
for lang in langs:
    with open(os.devnull, "w") as f, redirect_stdout(f):
        splits = get_splits(lang)
        train = splits["projz_train_path"]
        dev = splits["projz_dev_path"]
        test = splits["projz_test_path"]
    print(f"Treebank: {lang}")
    for split, split_name in [(train, "train"), (dev, "dev"), (test, "test")]:
        print(f"Split: {split_name}")
        all_unique_deprels, deprel_count = count_deprels(split)
        print(f"All unique dependency relations: {all_unique_deprels}")
        print(f"Count of all unique dependecy relations: {deprel_count}")
    print("---------------------------------------------------")
```
